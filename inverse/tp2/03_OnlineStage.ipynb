```python
import sys
from pyprojroot import here

# Ritorna il percorso assoluto della root del progetto
PROJECT_ROOT = str(here()) + "/"
sys.path.append(PROJECT_ROOT)
sys.dont_write_bytecode = True

print(f"La root del progetto è: {PROJECT_ROOT}")
```

```python
from paths import INV_PATH, TP2_PATH, MODEL_PATH, COLUMN_PATH

ABS_PATH = PROJECT_ROOT + INV_PATH + TP2_PATH
LOAD_MODEL = PROJECT_ROOT + MODEL_PATH + COLUMN_PATH

model_name = "column"
xdmf_file_name = "column.xdmf"

load_params = ABS_PATH + "files/pred_parameters.csv"

dbdir = ABS_PATH + "database/"
```

```python
import numpy as np
import fenics as fe
import pandas as pd

from modelaquisition.msh2xdmf import Msh2Xdmf

from fem_problems.invTP2.finite_element import SystemParabolicFEM
from fem_problems.invTP2.rbnics_pod_temp import TemporalPOD
```

```python
df = pd.read_csv(load_params, sep=";", index_col=0)

mu_pred = [df.iloc[i, 0] for i in range(3)]
```

```python
# Path to the XDMF file
file_path = ABS_PATH + xdmf_file_name

# Carica la mesh da file XDMF
mesh = fe.Mesh()
with fe.XDMFFile(file_path) as infile:
    infile.read(mesh)
```

```python
mu_range = [
    (0, 1.),
    (0, 1.),
    (0, 1.),
]
```

```python
temporal_steps = 21
time_interval = [0, 1]

fem_p = SystemParabolicFEM(mesh, time_interval, temporal_steps)
pod = TemporalPOD(mu_range, fem_p)
```

```python
pod.load_reduction(directory=dbdir, filename=model_name)
```

```python
mu_test = [.1, .2, .5]
mat, _ = pod.solve_online_rom(mu_test, pod.num_basis)
fom_m, _ = pod.fem_p.solve_online_fem(mu_test)
real_m, _ = pod.fem_p.exact_solution_online(mu_pred)

print("ERRORE FOM-ESATTA")
e, f = pod.fem_p.compute_error(fom_m, real_m)
print("ERRORE ROM-ESATTA")
c, d = pod.fem_p.compute_error(mat, real_m)
print("ERRORE ROM-FOM")
a, b = pod.fem_p.compute_error(mat, fom_m)
```

```python
rock = Msh2Xdmf(ABS_PATH + xdmf_file_name, model_name)

rock.reset_files("_sol_rom")
rock.reset_files("_sol_fom")
rock.reset_files("_sol_an")

rock.add_solution(mat, "_sol_rom", "solution_rom", steps=pod.fem_p.temp_points)
rock.add_solution(fom_m, "_sol_fom", "solution_fom", steps=pod.fem_p.temp_points)
rock.add_solution(real_m, "_sol_an", "solution_an", steps=pod.fem_p.temp_points)
```

```python
err = [np.abs(mat[i] - real_m[i]) / np.linalg.norm(real_m[i]) for i in range(len(mat))]

rock.reset_files("_err_rom_an")
rock.add_solution(err, "_err_rom_an", "error", steps=pod.fem_p.temp_points)
```

```python
err_ass_fom_rom = [np.abs(mat[i] - fom_m[i]) for i in range(len(mat))]
err_rel_fom_rom = [err_ass_fom_rom[i] / np.linalg.norm(fom_m[i]) for i in range(len(mat))]

rock.reset_files("_err_fom_rom")
rock.reset_files("_err_rel_fom_rom")
rock.add_solution(err_ass_fom_rom, "_err_fom_rom", "error_fom_rom", steps=pod.fem_p.temp_points)
rock.add_solution(err_rel_fom_rom, "_err_rel_fom_rom", "error_rel_fom_rom", steps=pod.fem_p.temp_points)
```

```python
err_fom_an = [np.abs(fom_m[i] - real_m[i]) for i in range(len(mat))]
err_rel_fom_an = [err_fom_an[i] / np.linalg.norm(real_m[i]) for i in range(len(mat))]

rock.reset_files("_err_fom")
rock.add_solution(err_rel_fom_an, "_err_fom", "error_fom", steps=pod.fem_p.temp_points)
```
