```python
import json
import torch
import fenics as fe

from pina import LabelTensor
from pina.model import ResidualFeedForward
```

```python
torch.set_default_dtype(torch.float64)
```

```python
# Apri il file JSON e caricalo in un dict
with open('file.json', 'r') as file:
    data_dict = json.load(file)

# Ora 'data_dict' è un dizionario Python con i dati del file JSON
print(data_dict)
```

```python
num_layers = data_dict['num_hidden_layers']
num_neurons = data_dict['num_hidden_neurons']
input_dim = data_dict['input_dim']
output_dim = data_dict['output_dim']
```

```python
model = ResidualFeedForward(input_dimensions=input_dim,output_dimensions=output_dim,n_layers=num_layers,inner_size=num_neurons)

# Caricare un modello salvato su GPU e utilizzarlo su CPU
device = torch.device('cpu')
model.load_state_dict(torch.load('model_weights.pth', map_location=device))
```

```python
file_path = "column.xdmf"

# Carica la mesh da file XDMF
mesh = fe.Mesh()
with fe.XDMFFile(file_path) as infile:
    infile.read(mesh)

mesh_points = LabelTensor(
    mesh.coordinates(),
    labels=['x', 'y', 'z']
)
```

```python
pred = model(mesh_points)
print(pred.tensor)
```
