```python
import sys
from pyprojroot import here

# Ritorna il percorso assoluto della root del progetto
PROJECT_ROOT = str(here()) + "/"
sys.path.append(PROJECT_ROOT)

print(f"La root del progetto è: {PROJECT_ROOT}")
```

```python
from paths import DRT_PATH, TP4_PATH

ABS_PATH = PROJECT_ROOT + DRT_PATH + TP4_PATH
LOAD_MSH = PROJECT_ROOT

model_name = "rock"
xdmf_file_name = "rock.xdmf"
data_path = ABS_PATH + "/files/TempData.csv"
```

```python
import pandas as pd
import fenics as fe

from fem_problems.dirTP4.finite_element import HeatFEM
```

```python
data = pd.read_csv(data_path, sep=";", index_col=0).values
data = data.reshape((data.shape[0],))
```

```python
# Path to the XDMF file
file_path = ABS_PATH + xdmf_file_name

# Carica la mesh da file XDMF
mesh = fe.Mesh()
with fe.XDMFFile(file_path) as infile:
    infile.read(mesh)
```

```python
fem = HeatFEM(mesh, [0, 1], data)
```

```python
sol, _ = fem.solve_fem()
```

```python
df = pd.DataFrame(
    data=sol,
)

df.to_csv(ABS_PATH + "/files/fem_solution.csv", sep=";", index=None, index_label=None)
```
