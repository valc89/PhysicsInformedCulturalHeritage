```python
import sys
from pyprojroot import here

# Ritorna il percorso assoluto della root del progetto
PROJECT_ROOT = str(here()) + "/"
sys.path.append(PROJECT_ROOT)

print(f"La root del progetto è: {PROJECT_ROOT}")
```

```python
from paths import DRT_PATH, TP4_PATH, MODEL_PATH, ROCK_PATH

ABS_PATH = PROJECT_ROOT + DRT_PATH + TP4_PATH
LOAD_MODEL = PROJECT_ROOT + MODEL_PATH + ROCK_PATH

model_name = "rock"
pinn_sol_path = ABS_PATH + "./files/pinn_solution.csv"
fem_sol_path = ABS_PATH + "./files/fem_solution.csv"

xdmf_file_name = "rock.xdmf"
```

```python
import numpy as np
import pandas as pd

from modelaquisition.msh2xdmf import Msh2Xdmf
```

```python
fem_sol = pd.read_csv(fem_sol_path, sep=";", index_col=False).values
pinn_sol = pd.read_csv(pinn_sol_path, sep=";", index_col=False).values
```

```python
err = np.abs(fem_sol-pinn_sol)
err_r = err/np.linalg.norm(fem_sol)
err_assoluto = np.linalg.norm(err)
err_relative = err_assoluto/np.linalg.norm(fem_sol)

print(f"Errore assoluto = {err_assoluto:.4e}\nErrore relativo = {err_relative:.4e}")
```

```python
rock = Msh2Xdmf(ABS_PATH + xdmf_file_name, model_name)

rock.reset_files("_sol_pinn")
rock.reset_files("_sol_fem")

rock.add_solution(pinn_sol.T, "_sol_pinn", "solution_pinn", steps=pinn_sol.shape[1])
rock.add_solution(fem_sol.T, "_sol_fem", "solution_fem", steps=fem_sol.shape[1])
```

```python
rock.reset_files("_err")
rock.reset_files("_err_rel")
rock.add_solution(err.T, "_err", "error", steps=err.shape[1])
rock.add_solution(err_r.T, "_err_rel", "error_rel", steps=err_r.shape[1])
```
